```python
import requests
import json
import os
```

```python
base_url = "http://localhost:5000"
```

## Creación de DB vectorial

```python
url = f"{base_url}/createDB"
nameFiles = os.listdir("documentos")

payload = {
        "model": "text-embedding-v3",
        "name": "CUSCATLAN",
        "nameRandom": False
    }

for i in nameFiles:
    with open(f"documentos/{i}", "rb") as file:
        response = requests.post(url, data=payload, files={"fileText": file})
        print("Response from /createDB:", response.json())
```

## Prueba con texto

```python
mess = [{"role": "user", "content": "tendrás descuentos en farmacias?"}]
url = f"{base_url}/assistant"
payload = {
    "history": mess,
    "nameDB": "CUSCATLAN",
}

# Realizar la solicitud POST
response = requests.post(url, json=payload)

# Imprimir la respuesta del servidor
print("Response from /assistant:", response.json())
```

```python
for i in response.json():
    print(response.json()[i])
```

## Prueba con imagen

```python
import base64

def image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

image_path = "images.jpg"
base64_image = image_to_base64(image_path)
```

```python
mess = [{"role": "user", "content": "tendrás uno o más descuentos de esto?"}]
url = f"{base_url}/assistant"
payload = {
    "history": mess,
    "nameDB": "CUSCATLAN_ERuz1GNfjnf4ZBdt.json",
    "image64": base64_image
}

# Realizar la solicitud POST
response = requests.post(url, json=payload)

# Imprimir la respuesta del servidor
print("Response from /assistant:", response.json())
```

```python
for i in response.json():
    print(response.json()[i])
```
